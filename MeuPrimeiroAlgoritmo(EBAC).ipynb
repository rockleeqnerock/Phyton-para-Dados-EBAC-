```python
#Adição (+) e Subtração (-)
print (5+2)
print (5-2)
```

#Variáveis

#Condicionais

```python
#As condicionais "if", "if/else" e "if/elif/else" servem para condicionar uma ação a algo.
#if:
idade_usuario = int(input("Digite a sua idade: "))

if idade_usuario >= 18:
    print("Acesso liberado.")
```

```python

#if/else:
idade_usuario = int(input("Digite a sua idade: "))

if idade_usuario >= 18:
    print("Acesso liberado.")
else:
    print("Acesso negado.")
```

```python
#if/elif/else:
idade_usuario = int(input("Digite a sua idade: "))

if idade_usuario >= 18:
    print("Acesso liberado.")
elif idade_usuario >= 16:
    print("Acesso liberado apenas com a supervisão dos pais.")
else:
    print("Acesso negado.")
```

#Operadores Logicos

```python
#Os operadores lógicos "and", "or" e "not" são uteis no uso de condicionais:
#and:
idade_usuario = int(input("Digite a sua idade: "))

if idade_usuario >= 18 and idade_usuario <= 60:
    print("Acesso liberado.")
else:
    print("Acesso negado.")
```

```python
#or:
idade_usuario = int(input("Digite a sua idade: "))

if idade_usuario >= 18 or idade_usuario <= 60:
    print("Acesso liberado.")
else:
    print("Acesso negado.")
```

```python
 #not:
idade_usuario = int(input("Digite a sua idade: "))

if not idade_usuario >= 18:
    print("Acesso negado.")
else:
    print("Acesso liberado.")
```

#Listas, Tuplas e Dicionário


```python
#Lista
Lista = [1,2,3,4,5]

#podemos adicionar elementos em uma lista:
Lista.append("Lee")
print(Lista)

#a lista tem posições predefinidas, podemos escolher uma posição para informar:
print(Lista[5])
```

```python
#Tupla
Tupla = (1,2,3,4,5)
print(Tupla)

#a tupla não pode ser alterada
```

```python
#Dicionário
Dicionário = {'num1': 1, 'num2': 2}
print(Dicionário)

#você adiciona valores a chaves em um dicionário:
print(Dicionário['num1'])

#para adicionar um valor a um dicionário não precisa utilizar nenhum comando:
Dicionário['num3'] = 3
print(Dicionário)
```

#Estruturas de Repetição

```python
# For range:

for i in range(1, 6):
    print(f"Número: {i}")
```

```python
#For em listas

frutas = ["Maçã", "Banana", "Laranja"]

for fruta in frutas:
    print(f"Eu gosto de {fruta}")
```

```python
#For percorrendo caracteres

palavra = "Python"

for letra in palavra:
    print(letra)
```

```python
#While

contador = 1

while contador <= 5:
    print(f"Contagem: {contador}")
    contador += 1
```

```python
#exemplo de while para entrada de usuário:

senha_correta = "python123"
tentativa = ""

# Continua pedindo a senha enquanto o usuário errar
while tentativa != senha_correta:
    tentativa = input("Digite a senha: ")

print("Acesso concedido!")
```
