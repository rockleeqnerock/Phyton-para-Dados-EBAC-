```python
operacao = input("selecione a operação: soma, subtração, multiplicação, potencia ou divisão")
num1 = int(input("insira o primeiro número: "))
num2 = int(input("insira o segundo número: "))

if operacao == "soma": print(num1 + num2)

if operacao == "subtração": print(num1 - num2)

if operacao == "multiplicação": print(num1 * num2)

if operacao == "divisão": print(num1 / num2)

if operacao == "potencia": print(num1 ** num2)

else: print("operação inválida")
```
