```python
print("=== CALCULADORA INTERATIVA ===")

while True:
    print("Operações disponíveis: soma, subtração, multiplicação, divisão, potencia")
    operacao = input("Selecione a operação (ou digite 'sair' para encerrar): ").strip().lower()

    if operacao == "sair":
        print("Encerrando a calculadora. Até mais!")
        break

    if operacao not in ["soma", "subtração", "multiplicação", "divisão", "potencia"]:
        print("Erro: Operação inválida. Por favor, escolha uma opção válida.")
        continue

    try:
        num1 = float(input("Insira o primeiro número: "))
        num2 = float(input("Insira o segundo número: "))
    except ValueError:
        print("Erro: Por favor, insira apenas números válidos.")
        continue

    if operacao == "soma":
        resultado = num1 + num2
        print(f"Resultado: {num1} + {num2} = {resultado}")

    elif operacao == "subtração":
        resultado = num1 - num2
        print(f"Resultado: {num1} - {num2} = {resultado}")

    elif operacao == "multiplicação":
        resultado = num1 * num2
        print(f"Resultado: {num1} * {num2} = {resultado}")

    elif operacao == "divisão":
        if num2 == 0:
            print("Erro: Não é possível realizar divisão por zero.")
        else:
            resultado = num1 / num2
            print(f"Resultado: {num1} / {num2} = {resultado}")

    elif operacao == "potencia":
        resultado = num1 ** num2
        print(f"Resultado: {num1} ** {num2} = {resultado}")
```
